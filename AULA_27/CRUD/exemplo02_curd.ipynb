```python
# APLICAÇÃO CRUD COM MYSQL.CONNECTOR

# Importando as bibliotecas
import mysql.connector

conexao = mysql.connector.connect(
    host = 'localhost',
    user = 'root',
    password = '123456',
    database = 'crud_2025'
)

cursor = conexao.cursor()
```

```python
# READ
comando = f"""
    SELECT * FROM tb_clientes
"""

cursor.execute(comando)
resultado = cursor.fetchall()

for r in resultado:
    print(r)
```

```python
# CREATE
nome = 'João'
idade = 25
telefone = '21 91111-1111'


comando = f"""
    INSERT INTO tb_clientes (NOME, IDADE, TELEFONE)
    VALUES ('{nome}', '{idade}', '{telefone}')
"""

# Realiza o comando
cursor.execute(comando)

# Persiste os dados no banco
conexao.commit()

# Fecha o cursor e a conexão com o banco
cursor.close()
conexao.close()

```

```python
# UPDATE
nome = 'Maria da Silva'
idade = 22
telefone = '21 98282-2525'

comando = f"""
    UPDATE tb_clientes
    SET NOME = '{nome}', 
        IDADE = '{idade}', 
        TELEFONE = '{telefone}'
    WHERE CODIGO = 2
"""

cursor.execute(comando)
conexao.commit()  # Atualiza os dados no banco

# Encerrando os recursos
cursor.close()
conexao.close()
```

```python
# DELETE
comando = f"""
    DELETE FROM tb_clientes
    WHERE CODIGO = 3
"""

cursor.execute(comando)
conexao.commit()  # Persiste as alterações no banco

cursor.close()
conexao.close()
```

```python
# read
with conexao.cursor() as cursor:
    comando = "SELECT * FROM tb_clientes"
    cursor.execute(comando)
    resultado = cursor.fetchall()
    for r in resultado:
        print(r)
```

```python
# CREATE
nome, idade, telefone = 'João', 25, '21 91111-1111'

with conexao.cursor() as cursor:
    comando = "INSERT INTO tb_clientes (NOME, IDADE, TELEFONE) VALUES (%s, %s, %s)"
    cursor.execute(comando, (nome, idade, telefone))
    conexao.commit() # O commit é na conexao, não no cursor
    print("Cliente inserido!")


# EXPLICAÇÃO
# comando = f"INSERT INTO tb_clientes (NOME) VALUES ('{nome}')"
# Usuário mal intencionado: 'João'); DROP TABLE tb_clientes; --'
# python entende:
# INSERT INTO tb_clientes (NOME) VALUES ('João'); DROP TABLE tb_clientes; --')

# explicação:
# O MySQL vê o primeiro ponto e vírgula (;) e entende que o INSERT acabou.
# Ele executa o próximo comando: DROP TABLE tb_clientes; (que apaga sua tabela inteira).
# O -- no final serve para o MySQL ignorar o restante da linha (aquela aspa que sobrou), evitando erro de sintaxe.

# Usando a sintaxe %s
# O resultado no banco seria:
# um cliente cadastrado com o nome esquisito: João'); DROP TABLE tb_clientes; --. 
# Nenhum comando é executado, apenas o texto é salvo.
```

```python
# UPDATE
nome, idade, telefone, codigo = 'Maria da Silva', 22, '21 98282-2525', 2

with conexao.cursor() as cursor:
    comando = "UPDATE tb_clientes SET NOME=%s, IDADE=%s, TELEFONE=%s WHERE CODIGO=%s"
    cursor.execute(comando, (nome, idade, telefone, codigo))
    conexao.commit()
    print(f"Registro {codigo} atualizado.")
```

```python
# DELETE
codigo_excluir = 3

with conexao.cursor() as cursor:
    comando = "DELETE FROM tb_clientes WHERE CODIGO = %s"
    cursor.execute(comando, (codigo_excluir,))
    conexao.commit()
    print(f"Registro {codigo_excluir} removido.")
```
